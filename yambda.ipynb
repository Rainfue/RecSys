# Загрузка данных

```python
path2data = 'Data/Yambda-50m'

dislikes = pd.read_parquet(os.path.join(path2data, 'dislikes.parquet'))
likes = pd.read_parquet(os.path.join(path2data, 'likes.parquet'))
listens = pd.read_parquet(os.path.join(path2data, 'listens.parquet'))
multi_event = pd.read_parquet(os.path.join(path2data, 'multi_event.parquet'))
undislikes = pd.read_parquet(os.path.join(path2data, 'undislikes.parquet'))
unlikes = pd.read_parquet(os.path.join(path2data, 'unlikes.parquet'))
```

# Описание данных

dislikes    |   пользовательские действия с дизлайками            | uid, item_id, timestamp, is_organic
  likes     |    пользовательские действия с лайками              | uid, item_id, timestamp, is_organic
listens     |   прослушивание с деталями воспроизведения          | uid, item_id, timestamp, is_organic, played_ratio_pct, track_length_seconds
undislikes  |   пользовательские действия с отменой дизлайков     | uid, item_id, timestamp, is_organic
unlikes     |   пользовательские действия с отменой лайков        | uid, item_id, timestamp, is_organic
